```python
ax=sns.countplot(x=df["Churn"],data=df)
ax.bar_label(ax.containers[0])
plt.title("Count of Customers Churned")
plt.show()
```

```python
gb = df.groupby("Churn").agg({'Churn':'count'})
plt.pie(gb['Churn'],labels = gb.index, autopct = "%1.2f%%")
plt.title("Percentage of Churned Customer")
plt.figure(figsize= (3,4))
plt.show()
```

```python


# Calculate percentage using groupby
pct = (
    df.groupby(['SeniorCitizen', 'Churn'])
      .size()
      .groupby(level=0)
      .apply(lambda x: x / x.sum() * 100)
      .unstack()
)

# 100% stacked bar chart
ax = pct.plot(
    kind='bar',
    stacked=True,
    figsize=(4,4)
)

plt.title("Churn by SeniorCitizen (%)")
plt.xlabel("SeniorCitizen")
plt.ylabel("Percentage (%)")
plt.ylim(0, 100)
plt.xticks(rotation=0)
plt.legend(title="Churn")

# Add percentage labels
for container in ax.containers:
    ax.bar_label(
        container,
        fmt='%.1f%%',
        label_type='center'
    )

plt.tight_layout()
plt.show()
```

```python
cols = [
    'PhoneService', 'MultipleLines', 'InternetService',
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
    'TechSupport', 'StreamingTV', 'StreamingMovies'
]
#create subplots
fig, axes = plt.subplots(3, 3, figsize=(15, 12))

for col, ax in zip(cols, axes.flat):
    sns.countplot(data=df, x=col,hue='Churn',ax=ax)
    ax.set_title(f'{col} vs Churn')
    ax.set_xlabel(col)
    ax.set_ylabel('Count')
    ax.tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()
```

#The countplots show that churn varies across different telecom services.
Customers using Fiber optic internet appear to have comparatively higher churn.
Customers without OnlineSecurity, OnlineBackup, DeviceProtection, and TechSupport generally show more churn.
StreamingTV and StreamingMovies show churn in both subscribed and non-subscribed groups.
Overall, internet service and additional support/security services appear to be important factors related to customer churn.
